```python
# 1. 冒險者登入與裝備選擇
name = input("請輸入你的冒險者名字：").strip()

print("\\n請選擇你的專屬裝備（每件裝備都有不同的戰鬥技能）：")
print("1. 【黃金鍵盤】技能「系統透視」：每回合有 50% 機率提前偷看到電腦要出的拳")
print("2. 【魔法魔杖】技能「幸運爆擊」：獲勝時有 50% 機率觸發爆擊，一口氣獲得 2 勝")
print("3. 【時光沙漏】技能「逆轉時光」：每場限用 1 次，戰敗時自動倒轉時間，強制變成平手")

# 確保玩家輸入正確的裝備編號
while True:
    choice = input("請輸入裝備編號 (1、2 或 3)：").strip()
    if choice == "1":
        equipment = "黃金鍵盤"
        break
    elif choice == "2":
        equipment = "魔法魔杖"
        break
    elif choice == "3":
        equipment = "時光沙漏"
        break
    else:
        print("⚠️ 指令錯誤，請輸入 1、2 或 3 來選擇裝備！\\n")

# 印出前置劇情
print("\\n=== 系統廣播 ===")
print(f"歡迎勇敢的冒險者【{name}】！")
print(f"你已經裝備了【{equipment}】，準備好迎接【猜拳】終極挑戰！")

print("\\n\\n=== ✊✋✌️ 剪刀石頭布 五局三勝大對決 ===")
print("規則：採用「五局三勝制」，先獲得 3 勝的一方就是最終贏家！（平手不計入勝場）\\n")

# 戰鬥前置作業：設定變數來記錄雙方的勝場數、回合數與裝備狀態
player_wins = 0
computer_wins = 0
round_num = 1
hourglass_used = False  # 記錄「時光沙漏」是否已經使用過

# 設定電腦可以出的拳
options = ["剪刀", "石頭", "布"]

# 2. 迴圈條件設定：只要雙方勝場都小於 3，比賽就繼續（五局三勝制）
while player_wins < 3 and computer_wins < 3:
    print(f"--- 第 {round_num} 回合 ---")
    print(f"🏆 目前戰況 - {name}:{player_wins} 勝 | 電腦：{computer_wins} 勝")

    # 電腦先在心中決定好要出的拳（這樣黃金鍵盤才能真正「預知」電腦的選擇）
    computer_choice = random.choice(options)
    # 【黃金鍵盤】被動技能觸發判定
    if equipment == "黃金鍵盤":
        if random.random() < 0.5:
            print(f"⌨️ 【黃金鍵盤】發動「系統透視」！偵測到電腦這回合準備出：【{computer_choice}】")
        else:
            print("⌨️ 【黃金鍵盤】訊號受到干擾，這回合無法偵測電腦的出拳。")

    # 3. 玩家出拳
    player_choice = input("請出拳 (輸入 剪刀、石頭 或 布)：").strip()

    # 檢查玩家是否輸入錯誤
    if player_choice not in options:
        print("⚠️ 哎呀！指令錯誤，請輸入 剪刀、石頭 或 布。\\n")
        continue  # 直接重新開始新的一回合

    print(f"🤖 電腦出了：{computer_choice}")

    # 4. 勝負判斷與裝備技能結算
    if player_choice == computer_choice:
        print("平手！沒有人得分。\\n")

    elif (player_choice == "剪刀" and computer_choice == "布") or \
         (player_choice == "石頭" and computer_choice == "剪刀") or \
         (player_choice == "布" and computer_choice == "石頭"):
        # 【魔法魔杖】被動技能觸發判定
        if equipment == "魔法魔杖" and random.random() < 0.5:
            print("🪄 【魔法魔杖】發動「幸運爆擊」！這回合直接獲得 2 勝！\\n")
            player_wins += 2
        else:
            print("🎉 這回合你贏了！\\n")
            player_wins += 1

    else:
        # 【時光沙漏】被動技能觸發判定
        if equipment == "時光沙漏" and not hourglass_used:
            hourglass_used = True
            print("⏳ 【時光沙漏】發動「逆轉時光」！時間倒流，這回合強制改判為平手！（沙漏能量已耗盡）\\n")
        else:
            print("💥 這回合電腦贏了！\\n")
            computer_wins += 1

    round_num += 1

# 5. 最終結算
print("=== 🌟 最終結果 🌟 ===")
# 注意：因為魔法魔杖爆擊可能會讓勝場變成 4 勝，所以這裡要用 >= 3 來判斷
if player_wins >= 3:
    print(f"🏆 恭喜【{name}】使用【{equipment}】以 {player_wins} : {computer_wins} 擊敗了電腦，贏得五局三勝大對決！")
else:
    print(f"💻 很可惜，電腦以 {computer_wins} : {player_wins} 獲得了最終勝利，下次再來挑戰吧！")
```
